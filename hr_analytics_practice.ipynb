```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\LENOVO\Desktop\csv_datasheets\hr_analytics_practice_dataset - Copy.csv")
```

Cleaning Data

```python
df.info()
```

```python
df.drop_duplicates()
df["Hire_Date"] = pd.to_datetime(df["Hire_Date"])
df["Left"] = df["Status"].isin(["Resigned", "Terminated"]).astype(int)
df["Hire_Year"] = df["Hire_Date"].dt.year
```

Making Plot

```python
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x="Status")
plt.title("Employee Status")
plt.show()
```

```python
plt.figure(figsize=(8, 4))
sns.countplot(data=df, y="Department")
plt.title("Employees by Department")
plt.show()
```

```python
plt.figure(figsize=(8, 4))
sns.barplot(data=df, x="Left", y="Department")
plt.title("Attrition Rate by Department")
plt.xlabel("Attrition rate (0 to 1)")
plt.show()
```

```python
plt.figure(figsize=(8, 4))
sns.barplot(data=df, x="Job_Level", y="Base_Salary")
plt.title("Average Base Salary by Job Level")
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.barplot(data=df, x="Status", y="Satisfaction_Score")
plt.title("Average Satisfaction by Status")
plt.show()
```
